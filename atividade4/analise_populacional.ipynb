# Análise Populacional - Censo 2010 e 2022

```python
import pandas as pd
import os
```

```python
caminho = r'\\fserverb\homes\Python\3° Etapa\Analise dados\atividade4\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx'
df = pd.read_excel(caminho, header=2)

df = df.drop(columns=['Unnamed: 0'])

df.columns = ['UF', 'COD_UF', 'COD_MUNIC', 'NOME_MUNICIPIO', 'POP_2010_SINOPSE', 'POP_2010_COMPAT', 'POP_2022']

print('Shape:', df.shape)
print('\nPrimeiras linhas:')
df.head()
```

```python
pop_estados = df.groupby('UF')[['POP_2010_COMPAT', 'POP_2022']].sum().reset_index()

pop_estados['CRESCIMENTO'] = pop_estados['POP_2022'] - pop_estados['POP_2010_COMPAT']

pop_estados = pop_estados.sort_values('CRESCIMENTO', ascending=False)

print('População por Estado (ordenada por crescimento):')
pop_estados
```

```python
caminho_saida_estados = r'\\fserverb\homes\Python\3° Etapa\Analise dados\atividade4\populacao_estados_crescimento.csv'
pop_estados.to_csv(caminho_saida_estados, sep=';', index=False)
print(f'Tabela de estados salva em: {caminho_saida_estados}')
```

```python
pop_municipios = df[['UF', 'COD_UF', 'COD_MUNIC', 'NOME_MUNICIPIO', 'POP_2010_COMPAT', 'POP_2022']].copy()
pop_municipios['CRESCIMENTO'] = pop_municipios['POP_2022'] - pop_municipios['POP_2010_COMPAT']

pop_municipios = pop_municipios.sort_values('CRESCIMENTO', ascending=False)

print('População por Município (ordenada por crescimento):')
pop_municipios.head(20)
```

```python
caminho_saida_municipios = r'\\fserverb\homes\Python\3° Etapa\Analise dados\atividade4\populacao_municipios_crescimento.csv'
pop_municipios.to_csv(caminho_saida_municipios, sep=';', index=False)
print(f'Tabela de municípios salva em: {caminho_saida_municipios}')
```
